```python
from langchain_community.document_loaders import UnstructuredWordDocumentLoader

file_path = r"C:\Users\Rose\Pictures\Screenshots\wow.docx"

print("Loading .DOC file with structure preservation...")

loader = UnstructuredWordDocumentLoader(
    file_path,
    mode="elements",  # Preserves document structure
    infer_table_structure=True,  # Preserve table structure
    extract_images=True,  # Extract image information
    include_page_breaks=True,  # Preserve page structure
    languages=["fa", "en"]  
)

documents = loader.load()

print(documents)  
```
